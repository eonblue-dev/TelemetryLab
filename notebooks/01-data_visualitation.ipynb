### 🏎️ FastF1 Session Setup

We initialize the session using `fastf1.get_session()`. Here is a breakdown of the parameters:

*   **`year`** *(int)*: The championship year (e.g., `2024`, `2026`).
*   **`gp`** *(str | int)*: The Grand Prix name (e.g., `'Monaco'`) or round number (e.g., `1`). Uses fuzzy matching by default.
*   **`identifier`** *(str | int)*: The specific event of the weekend:
    *   *Abbreviations:* `'FP1'`, `'FP2'`, `'FP3'`, `'Q'`, `'S'`, `'SQ'`, `'R'` (Race).
    *   *Session numbers:* `1` to `5` in chronological order.
*   **`backend`** *(str, optional)*: API data source (`'fastf1'`, `'f1timing'`, `'jolpica-f1'`). *Keyword-only.*
*   **`exact_match`** *(bool, optional)*: If `True`, disables fuzzy matching for the GP name. *Keyword-only.*

> **Note:** `get_session()` only instantiates the object. You must call `.load()` afterwards to fetch telemetry, laps, and timing data.

```python
import fastf1
fastf1.Cache.enable_cache('../cache/')
session = fastf1.get_session(2025,'Spain','R')
session.load()
```
